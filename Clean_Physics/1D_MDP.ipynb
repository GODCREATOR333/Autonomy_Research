A **Dichotomous Markov Process (DMP)** is a stochastic process where an agent switches between **two discrete states** with given probabilities per unit time: **G state**: Geocentric / directed motion (drift with velocity $v$), **E state**: Egocentric / exploratory motion (diffusion with coefficient $D$). The switching rates are: $\alpha$ → probability per unit time to switch from G → E, $\beta$ → probability per unit time to switch from E → G.  

The **Mean First Passage Time (MFPT)** is the expected time for an agent starting at position $x=0$ to reach a goal at $x=L$ for the **first time**. For a DMP in 1D, the **backward equations** for MFPT are:  
$$
\begin{cases}
v \frac{d T_G}{dx} + \alpha (T_E - T_G) = -1 \\
D \frac{d^2 T_E}{dx^2} + \beta (T_G - T_E) = -1
\end{cases}
$$
where $T_G(x)$ → MFPT starting in G state at position $x$, $T_E(x)$ → MFPT starting in E state at position $x$, with boundary conditions $x=0$ reflecting, $x=L$ absorbing.  

Applications in real life include: Cellular transport: Proteins/vesicles alternate between active transport along filaments and diffusion, MFPT predicts delivery time to target organelles; Animal foraging / search behavior: Animals alternate between directed movement and exploratory searching, MFPT gives expected time to find food or a mate; Robotics / autonomous exploration: Robots switch between fast movement toward a target and slow scanning, optimizing switching reduces time to reach goals; Physics / chemistry: Diffusion-limited reactions, escape from traps, or stochastic transport, helps understand how intermittent dynamics affect reaction or escape times.  

**Summary**: The DMP + MFPT framework is a **minimal model** to study intermittent search or transport strategies, and can be generalized to more complex dynamics and environments.
