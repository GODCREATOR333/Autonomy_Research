# Day 2 — EGO Theory Only

## 2. Analyze the Egocentric policy as an ABP / persistent random walk
### Objective
Understand the EGO policy rigorously as a **physics object**.

### Formalize
- State:
  \[
  s_t = (x_t, a_{t-1}, w_t)
  \]
  where:
  - \(x_t\) = position
  - \(a_{t-1}\) = previous action / heading
  - \(w_t\) = local wall observation

- Transition kernel:
  \[
  P_{\text{ego}}(s_{t+1} \mid s_t)
  \]

### Compute / Derive
- Directional autocorrelation
- Persistence length / persistence time
- Entropy of the action distribution
- Mean displacement / effective diffusion behavior
- Expected coverage / unique cells visited
- First-passage / hitting-time intuition
- Why collisions are ~0 by construction

### Deliverable
Write a short technical note:

## Title
**EGO as a locally constrained persistent random walk**

### Objective

### Formalize

\[
\pi(a \mid x) \propto \exp(\kappa \, \hat{u}_g(x) \cdot \hat{a})

State:
\[
s_t = x_t

P_{\text{geo}}(x_{t+1} \mid x_t)

# After EGO + GEO Are Clean

# 4. Start the switching theory

## 4.1 Define the switching variable
Introduce a mode variable:
\[
m_t \in \{\text{ego}, \text{geo}\}
\]

This means the agent can switch between:
- local obstacle-aware exploration
- global goal-directed drift

---

## 4.2 Write the full switching process
Formalize the joint dynamics:
\[
P(s_{t+1}, m_{t+1} \mid s_t, m_t)
\]

This is the beginning of the **dichotomous Markov process / master equation** formulation.

---

## 4.3 Define the optimization target
The real objective is:
\[
\min_{\{m_t\}} \mathbb{E}[\tau_{\text{goal}}]
\]

Possible variants:
- Minimize expected time-to-goal
- Maximize success probability under time limit
- Minimize collisions + time jointly
